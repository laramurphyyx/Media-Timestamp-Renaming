```python
from pathlib import Path
import os
import shutil

folder = Path(r"D:\lara\Snapchat with Exif")
missingsuffix_folder = "D:/lara/Snapchat with Exif/missingsuffix/"

for file in folder.iterdir():
    if file.is_file():
        if " - Snapchat" not in file.stem:
            
            new_name = file.stem + " - Snapchat" + file.suffix
            new_path = file.with_name(new_name)
            
            if Path(new_path).exists():
                missingsuffix_path = missingsuffix_folder + new_name
                shutil.copyfile(str(file), missingsuffix_path)
                if Path(missingsuffix_path).exists():
                    os.remove(str(file))
```

```python
from pathlib import Path
import os
import shutil

folder = Path(r"D:\lara\Snapchat with Exif")
main_folder = "D:/lara/Snapchat with Exif/mains/"
overlay_folder = "D:/lara/Snapchat with Exif/overlays/"

for file in folder.iterdir():
    if file.is_file():
        if "main" in file.stem:
            new_path = main_folder + file.name
            shutil.copyfile(str(file), new_path)
          
            if Path(new_path).exists():
                os.remove(str(file))
                
        elif "overlay" in file.stem:
            new_path = overlay_folder + file.name
            shutil.copyfile(str(file), new_path)
          
            if Path(new_path).exists():
                os.remove(str(file))
```
