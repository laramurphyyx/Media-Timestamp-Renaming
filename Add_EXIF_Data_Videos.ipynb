```python
import subprocess

def set_video_datetime(path, dt):
    # dt = "2018-01-01T12:01:05"
    
    subprocess.run([
        "exiftool",
        f"-creation_time={dt}",
        f"-MediaCreateDate={dt.replace('-', ':').replace('T', ' ')}",
        f"-CreateDate={dt.replace('-', ':').replace('T', ' ')}",
        "-overwrite_original",
        path
    ])
```

```python
EXIFTOOL_PATH = r"C:\Program Files\Exiftools\exiftool-13.55_64\exiftool.exe"
```

```python
import subprocess
import json

def get_video_metadata(path):
    result = subprocess.run(
        [EXIFTOOL_PATH, "-j", path],
        capture_output=True,
        text=True
    )
    return json.loads(result.stdout)[0]
```

```python
import subprocess
from pathlib import Path

def overlay_video(main_path, overlay_path, output_path):

    cmd = [
        "ffmpeg",
        "-i", main_path,
        "-i", overlay_path,
        "-filter_complex", "overlay=0:0",
        "-c:v", "libx264",
        "-c:a", "copy",
        "-map_metadata", "0",
        output_path
    ]

    subprocess.run(cmd, check=True)

    if Path(output_path).exists():
        print("Video overlay successful:", output_path)
```

```python
overlay_video(main_path, overlay_path, output_path)
```
