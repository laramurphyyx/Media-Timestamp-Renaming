```python
from PIL import Image
from pathlib import Path
import os
```

```python
def overlay_image(main_path, overlay_path, output_path):
    
    # Open images
    base = Image.open(main_path).convert("RGBA")
    exif = base.info.get("exif")
    overlay = Image.open(overlay_path).convert("RGBA")

    # Ensure same size (Snapchat overlays usually match, but just in case)
    overlay = overlay.resize(base.size)

    # Combine
    combined = Image.alpha_composite(base, overlay)

    # Save as JPG (convert back to RGB because JPG doesn't support alpha)
    combined.convert("RGB").save(output_path, "JPEG", exif=exif)
    
    if Path(output_path).exists():
        print("Copy successful: " + output_path)
        os.remove(overlay_path)
```

```python
overlay_folder = "C:/Users/laram/Pictures/20170223 - 2019 Renamed Snapchats/"
main_folder = "C:/Users/laram/Pictures/20170223 - 2019 Renamed Snapchats with EXIF/"
```

```python
for file in Path(overlay_folder).iterdir():
    if file.is_file():
        
        if 'overlay' not in file.stem:
            continue

        overlay_file_path = str(file)
        overlay_filename = file.stem
        main_filename = overlay_filename.replace("_overlay", "_main")
        main_file_path = ""
        
        for suffix in ['.jpg', '.jpeg', '.JPG', '.JPEG']:
            
            test_main_path = main_folder + main_filename + suffix
            
            
            if Path(test_main_path).exists():
                main_file_path = test_main_path
                main_file_suffix = Path(main_file_path).suffix
                
#             else:
#                 print("Couldn't find:", test_main_path)
        
        if main_file_path:
            
            print("Found main file: " + main_file_path)
            overlay_output_file_path = main_folder + overlay_filename + main_file_suffix
            print("Saving to: " + overlay_output_file_path)
            
            overlay_image(main_file_path, overlay_file_path, overlay_output_file_path)
```
